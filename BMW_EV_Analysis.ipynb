### Source:
Adapted from (BMW Report 2024,p.62)

link:https://www.bmwgroup.com/en/report/2024/downloads/index.html

```python
Data2=pd.DataFrame({ 
    'Plants':['Spartanburg(USA)','Dadong(China)','Regensburg(Germany)',
              'Dingolfing(Germany)','Tiexi(China)','Leipzig(Germany)',
              'Munich(Germany)','Oxford(UK)','San Luis Potosi(Mexico)',
              'Rosslyn(South Africa)','Chennai(India)','Araquari(Brazil)',
              'Rayong(Thailand)','Goodwood(UK)','Debrecen(Hungary)',
              'Zhangjiagang(China)','Born-VDL Nedcar(Netherlands)',
              'GrazMagna Steyr(Austria)','Partner Plants'],
    '2024':[396117,343973,342521,297761,284045,246195,200590,
            110939,95236,55516,14568,11472,8666,5924,92,67561,
            7515,10463,14676],
    '2023':[410793,420586,238301,291907,307972,188199,217480,
            185400,117576,68238,15264,10608,13044,6179,'',
            2871,120235,26461,20808]})
Data2.sample(1)
```

```python
Data2=Data2.sort_values(by='2024',ascending=True)
colors=['#595959']*(len(Data2)-3)+['#0066b3']*1+['Gray']*2
yfontcolor=['black']*(len(Data2)-3)+['#0066b3']*1+['black']*2
styled_ticks=[f"<span style='color:{color}'><b>{name}</b></span>"if color != 'black'else name for name,color in zip(Data2['Plants'],yfontcolor)]
bar_labels1=[f"{val/1000:.0f}k"if name=="Regensburg(Germany)" else "" for name, val in zip(Data2['Plants'],Data2['2024'])]
bar_labels2=[f"{val/1000:.0f}k"if name=="Regensburg(Germany)" else "" for name, val in zip(Data2['Plants'],Data2['2023'])]
fig=go.Figure()
fig.add_trace(go.Bar(
    y=Data2['Plants'],
    x=Data2['2024'],
    name='2024',
    text=bar_labels1,
    orientation='h',
    marker_color=colors,
    opacity=0.5
))
fig.add_trace(go.Bar(
    y=Data2['Plants'],
    x=Data2['2023'],
    name='2023',
    text=bar_labels2,
    textposition='outside',
    orientation='h',
    marker_color='grey',
    opacity=0.6
))
fig.update_layout(
    title=dict(text='''<span style='font size:20px;color:rgb(102,102,102);'><b> Identifying High Efficiency
Growth Hubs(<span style='color:#595959;'>2023</span> vs <span style='color:#0066b3;'>2024</span>)</b></span><br>
<sup>Most Plants Slowed Down While Key German Hubs Broke Records</sup>''',
              font_size=25,
              font_color='rgb(102,102,102)'),
    barmode='overlay',
    template='plotly_white',
    height=600,width=1000,
    showlegend=False,
    xaxis=dict(
        showgrid=False,
        showticklabels=False,
    ),
    yaxis=dict(
    tickmode='array',
    tickvals=list(range(len(Data2))),
    ticktext=styled_ticks,
        color='black'
    ),margin=dict(t=130,r=150,l=0,b=0))
fig.add_annotation(
    text="Now it's <b>Global Mandate,<br>need to accelerate investment</b><br>in high volume hubs to <b>prevent<br>market erosion.</b>",
    xref='paper',
    yref='paper',
    x=1.054,y=0.57,
    showarrow=False,
    align='left',
    font_color='grey'
)
fig.add_annotation(
    text='''<span style='color:rgb(128,177,211);'><b>Regensburg</b></span> proves that<br>flexibility in production
is<br>paying off,meeting<b><span style='color:rgb(128,177,211);'> electric<br>demand.</span>''',
    xref='x',
    yref='y',
    x=405000,y=13.9,
    showarrow=False,
    align='left',
    font_color='grey'
)
fig.add_annotation(
    xref='paper',
    yref='paper',
    text='in 1000 Units',
    x=-0.18,y=1.08,
    showarrow=False,
    font=dict(size=15,color='gray')
)
fig.add_shape(
    type='line',
    xref='paper',
    yref='paper',
    x0=-0.21,x1=1,
    y0=1.082,y1=1.082,
    line=dict(color='black',width=0.9)
)
fig.show()
```

# The EV Transition-BEV vs Total Market
### Source:
BMW Report 2024,p.81

link:https://www.bmwgroup.com/en/report/2024/downloads/index.html

```python
df=pd.DataFrame({
    'Sale_varient':['BMW Total','BEV','PHEV'],
    '2024':[2200217,368475,164172],
    '2023':[2252793,330197,173878],
    '%_change':[-2.3,11.6,-5.6]
})
df.sample(1)
```

```python
fig=go.Figure()
for i,row in df.iterrows():
    if row['Sale_varient']=='BMW Total':
        shrink_factor=3
    else:
        shrink_factor=0.1
    color='#0066b3' if row['Sale_varient']=='BEV' else '#949494' 
    y_plot=[row['2023']/shrink_factor,row['2024']/shrink_factor]
    fig.add_trace(go.Scatter(x=['2023','2024'],
                            y=y_plot,
                            mode='lines+markers+text',
                            line=dict(color=color,
                                     width=4),
                            marker=dict(size=10,color=color),
                            text=[f"{int(row['2023']):,}",f"{int(row['2024']):,}  {row['Sale_varient']}"],
                             textfont=dict(color=color),
                            textposition=["middle left","middle right"],
                            hoverinfo="skip"))
    fig.add_annotation(
        x=0.5,
        y=(y_plot[0]+y_plot[1])/2,
        text=f"{row['%_change']}%",
        showarrow=False,
        font=dict(color=color,
                 size=14,family='Arial Black'),
        yshift=20 if
        row['Sale_varient']=='BEV' else -20
    )
    fig.add_annotation(
        text="# OF UNITS",
        font=dict(color='#949494',size=12),
        xref='paper',
        yref='paper',
        x=0.09,
        y=0.65,
        showarrow=False
    )
    fig.add_annotation(
        text="% CHANGE",
        font=dict(color='#949494',size=12),
        xref='paper',
        yref='paper',
        x=0.4,
        y=0.65,
        showarrow=False
    )
    fig.add_annotation(
        text="<span style='color:#0066b3;'><b>Electrification</b></span> is the Future",
        font=dict(size=24,color='rgb(102,102,102)',family='Arial'),
        xref='paper',
        yref='paper',
        x=0.074,
        y=0.95,
        showarrow=False
    )
    fig.add_annotation(
        text="It's <b>Risky to Invest</b> in Traditional Segments",
        font=dict(size=22,color='rgb(102,102,102)',family='Arial'),
        xref='paper',
        yref='paper',
        x=0.074,
        y=0.89,
        showarrow=False
    )
    fig.add_annotation(
        text='''<b>Total deliveries</b> are contracting despite an<span style='color:#0066b3;'> 11.6%
surge in BEVs</span>,<br>highlighting the risk in legacy segments''',
        font=dict(size=13,color='gray',family='Arial'),
        xref='paper',
        yref='paper',
        align='left',
        x=0.074,
        y=0.82,
        showarrow=False
    )
    fig.update_layout(
        plot_bgcolor='white',
        showlegend=False,
        xaxis=dict(showgrid=False,
                  showline=False,
                   color='#949494',
                  range=[-0.5,2]),
        yaxis=dict(showgrid=False,
                  showticklabels=False,
                  range=[0,7000000]),
        margin=dict(l=0,r=0,t=0,b=30),
        height=400,width=600
    )
fig.show()
```

## Visual Highlights
Used Strategic color(Blue for BEV and Muted gray).Instantly directs eye to the BEV.

Direct labeling on axis remove the need of legend.

Remove x,y-axis to follow clutter is enemy principle.
### Insight
The chart explain that BEV growth is not only extra sales,but it is replacing traditional sales.The business must focus 100% of its innovation on BEVs to stabilize the Total line.
